```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from sklearn.preprocessing import StandardScaler

X = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [8, 7],
    [9, 8],
    [10, 9]
])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


single_link = linkage(X_scaled, method='single')

plt.figure(figsize=(8, 4))
dendrogram(single_link)
plt.title("Single Linkage")
plt.show()

single_clusters = fcluster(single_link, 2, criterion='maxclust')
print("Single Linkage Clusters:", single_clusters)


complete_link = linkage(X_scaled, method='complete')

plt.figure(figsize=(8, 4))
dendrogram(complete_link)
plt.title("Complete Linkage")
plt.show()

complete_clusters = fcluster(complete_link, 2, criterion='maxclust')
print("Complete Linkage Clusters:", complete_clusters)

average_link = linkage(X_scaled, method='average')

plt.figure(figsize=(8, 4))
dendrogram(average_link)
plt.title("Average Linkage")
plt.show()

average_clusters = fcluster(average_link, 2, criterion='maxclust')
print("Average Linkage Clusters:", average_clusters)
```
